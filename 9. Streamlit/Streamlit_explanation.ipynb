# What is Streamlit ?

- Streamlit is a Python framework that allows you to build simple and interactive web apps using only Python code. You don’t need HTML, CSS, or JavaScript. Think of it like this:

- You write Python → Streamlit turns it into a website instantly.

It is mainly used for:
- Showing data
- Creating dashboards
- Interactive UI (buttons, sliders, inputs)
- Demonstrating ML/AI projects
- Rapid prototyping


# Why Streamlit Is Useful ?

1. Fast Prototyping
- You can build a working web app in minutes.

2. Pure Python
- No frontend knowledge required everything is written in Python.

3. Live Updating
- Whenever users interact (click, upload, slide), Streamlit runs your script again to update the page.

4. Interactive Components: Streamlit provides ready-made UI elements like:
- Text
- Buttons
- Sliders
- Dropdowns
- File upload
- Graphs
- Layouts

5. Ideal for ML Projects
Although we are not doing ML yet, Streamlit is commonly used later to:
- Show inputs required for prediction
- Display graphs
- Host ML models as simple apps

# How Streamlit Works Internally

Streamlit reads your Python script from top to bottom and displays whatever UI elements you write.

Each time the user interacts:
- Script runs again
- New output is shown

This makes the app feel dynamic even though it's just python.

```python
import streamlit as st

st.title("Hello Streamlit")
st.write("This is my first Streamlit app!")
```
